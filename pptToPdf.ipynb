```python
import os
import time
from comtypes.client import CreateObject
from tqdm import tqdm  # 用來顯示漂亮的進度條

def batch_ppt_to_pdf(source_dir, output_dir):
    """
    將來源資料夾內的所有 PPT/PPTX 檔案批次轉換為 PDF 並存入目標資料夾
    """
    # 確保路徑正確並建立目標資料夾
    source_dir = os.path.abspath(source_dir)
    output_dir = os.path.abspath(output_dir)
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        print(f"已建立目標資料夾：{output_dir}")

    # 篩選出所有 PPT 和 PPTX 檔案
    files = [f for f in os.listdir(source_dir) if f.lower().endswith(('.ppt', '.pptx'))]
    
    if not files:
        print("❌ 來源資料夾中沒有找到任何 PPT 或 PPTX 檔案。")
        return

    print(f"📂 找到 {len(files)} 個簡報檔案，準備開始轉檔...")

    # 初始化 PowerPoint 後台程序
    powerpoint = CreateObject("PowerPoint.Application")
    # 1 代表在後台開啟（或可用 powerpoint.WindowState = 2 最小化）
    powerpoint.Visible = 1 

    success_count = 0
    fail_count = 0

    # 使用 tqdm 顯示進度條
    for file_name in tqdm(files, desc="轉檔進度", unit="檔"):
        ppt_path = os.path.join(source_dir, file_name)
        
        # 產生對應的 PDF 檔名
        base_name = os.path.splitext(file_name)[0]
        pdf_name = f"{base_name}.pdf"
        pdf_path = os.path.join(output_dir, pdf_name)

        try:
            # 開啟簡報（ReadOnly=True, Untitled=WithWindow=False 防止干擾）
            deck = powerpoint.Presentations.Open(ppt_path, WithWindow=False)
            
            # 32 代表 ppSaveAsPDF 格式
            deck.SaveAs(pdf_path, 32)
            deck.Close()
            success_count += 1
        except Exception as e:
            print(f"\n❌ 檔案 {file_name} 轉檔失敗。錯誤原因: {e}")
            fail_count += 1
            
    # 關閉 PowerPoint 應用程式
    powerpoint.Quit()
    print(f"\n🎉 任務完成！成功：{success_count} 檔 | 失敗：{fail_count} 檔")

# ==================== 使用範例 ====================
```

```python
if __name__ == "__main__":
    # 🛠️ 請在此處替換成你的資料夾路徑
    SOURCE_FOLDER = r"D:\Users\User\Downloads\HW\4-1 逆向\ppt"
    OUTPUT_FOLDER = r"D:\Users\User\Downloads\HW\4-1 逆向\pdf"
    
    batch_ppt_to_pdf(SOURCE_FOLDER, OUTPUT_FOLDER)
```
