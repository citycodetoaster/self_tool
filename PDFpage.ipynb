```python
import os
from pypdf import PdfReader, PdfWriter
```

```python
def split_pdf(file_path, start_page, end_page):
  # 取得檔案所在的資料夾路徑與完整檔名
  folder_dir = os.path.dirname(file_path)
  file_name = os.path.basename(file_path)

  # 分離「主檔名」與「副檔名 (.pdf)」
  base_name, ext = os.path.splitext(file_name)

  # 組合新的檔名與完整輸出路徑
  output_name = f"{base_name}_{start_page}_{end_page}{ext}"
  output_path = os.path.join(folder_dir, output_name)

  # 讀取與寫入 PDF
  reader = PdfReader(file_path)
  writer = PdfWriter()

  # 檢查頁碼範圍是否合理
  total_pages = len(reader.pages)
  if start_page < 1 or end_page > total_pages or start_page > end_page:
    print(f"錯誤：頁碼範圍無效。此檔案共有 {total_pages} 頁。")
    return

  # 擷取指定頁面（索引從 0 開始）
  for page_num in range(start_page - 1, end_page):
    writer.add_page(reader.pages[page_num])

  with open(output_path, "wb") as output_file:
    writer.write(output_file)

  print(f"拆分完成！檔案已儲存至：{output_path}")
```

```python
# --- 執行範例 ---
file_path = r"D:\Users\User\Downloads\HW\4-1 密碼學\2024-2076.pdf"
split_pdf(file_path, 1, 10)
```
